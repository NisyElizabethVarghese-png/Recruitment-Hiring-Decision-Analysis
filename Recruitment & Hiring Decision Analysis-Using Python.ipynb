<h3 style="border-left:5px solid #AD1457; padding-left:10px; color:#AD1457; font-weight:800;">
2.2.4 Outlier Detection Using IQR Method
</h3>

<p style="line-height:1.6; color:#333;">
The IQR method is used to identify values that are unusually low or high compared with the rest of the data.
</p>

<p style="line-height:1.6; color:#333;">
This check helps determine whether the numerical columns contain potential outliers before continuing with the analysis.
</p>

```python
numeric_columns = [
    "Age",
    "Experience_Years",
    "Skill_Score",
    "Aptitude_Test_Score",
    "Technical_Test_Score",
    "Communication_Score",
    "Certifications_Count",
    "Previous_Companies",
    "Interview_Score",
    "Expected_Salary"
]

outlier_counts = {}

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    outlier_counts[column] = len(outliers)

outlier_counts
```

```python
outlier_counts_df = pd.DataFrame(
    list(outlier_counts.items()),
    columns=["Column", "Outlier_Count"]
)

outlier_counts_df
```

<h4 style="color:#8E5CA8; margin-left:10px; font-weight:700;">
2.4.2 Creating Experience Category
</h4>

<p style="line-height:1.6; color:#333;">
The <strong>Experience_Years</strong> column is grouped into four categories to make it easier to compare candidates based on their level of experience.
</p>

```python
def categorize_experience(years):
    if years <= 2:
        return "Entry"
    elif years <= 5:
        return "Junior"
    elif years <= 10:
        return "Mid"
    else:
        return "Senior"

df["Experience_Category"] = df["Experience_Years"].apply(categorize_experience)
```

```python


df[["Experience_Years", "Experience_Category"]].head()
```

```python
df["Experience_Category"].value_counts()
```

```python

import numpy as np

df["Experience_Category"] = pd.cut(
    df["Experience_Years"],
    bins=[-np.inf, 1, 3, 5, np.inf],
    labels=["Entry", "Junior", "Mid", "Senior"]
)

print(df["Experience_Category"].value_counts().sort_index())
```

### Observation

* The candidates are grouped into four experience categories. 
* The Senior category has the highest number of candidates with 54,133, followed by Mid with 29,819, Junior with 18,075, and Entry with 17,974.

4.1.4 Job Role Distribution — Pareto Chart 
</h3>

<p style="line-height: 1.7; color: #444;">
This Pareto chart ranks job roles according to candidate counts and displays the cumulative percentage. It helps compare the number of applicants across roles and understand how the candidate pool is distributed among them.
</p>


```python



role_counts = (
    df["Job_Role_Applied"]
    .value_counts()
    .sort_values(ascending=False)
    .reset_index()
)

role_counts.columns = ["Job_Role", "Candidate_Count"]

role_counts["Cumulative_Percentage"] = (
    role_counts["Candidate_Count"].cumsum()
    / role_counts["Candidate_Count"].sum()
    * 100
)

fig = go.Figure()

fig.add_trace(go.Bar(
    x=role_counts["Job_Role"],
    y=role_counts["Candidate_Count"],
    name="Candidate Count",
    marker_color="#9575CD",
    text=role_counts["Candidate_Count"].map(lambda x: f"{x:,}"),
    textposition="outside",
    hovertemplate="Job Role: %{x}<br>Candidates: %{y:,}<extra></extra>"
))

fig.add_trace(go.Scatter(
    x=role_counts["Job_Role"],
    y=role_counts["Cumulative_Percentage"],
    name="Cumulative Percentage",
    mode="lines+markers+text",
    text=role_counts["Cumulative_Percentage"].map(lambda x: f"{x:.1f}%"),
    textposition="top center",
    yaxis="y2",
    line=dict(color="#E6A4B4", width=3),
    marker=dict(size=8),
    hovertemplate="Job Role: %{x}<br>Cumulative: %{y:.2f}%<extra></extra>"
))

fig.update_layout(
    title=dict(
        text="Pareto Analysis of Candidate Distribution by Job Role",
        x=0.5,
        font=dict(size=21, color="#5B2C83")
    ),
    xaxis_title="Job Role",
    yaxis=dict(title="Number of Candidates"),
    yaxis2=dict(
        title="Cumulative Percentage",
        overlaying="y",
        side="right",
        range=[0, 115],
        ticksuffix="%"
    ),
    template="plotly_white",
    legend=dict(orientation="h", y=1.12, x=0.5, xanchor="center")
)

fig.show()
```
